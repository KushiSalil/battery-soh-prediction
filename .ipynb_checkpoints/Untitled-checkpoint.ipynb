```python
import argparse, numpy as np
import torch.nn as nn
from dataloader.XJTU_loader import XJTUDdataset
from nets.Model import SOHMode

SOHMode._preprocessing_net = lambda self: nn.Linear(67, 128*4)

args = argparse.Namespace(
    random_seed=2023, data='XJTU',
    input_type='handcraft_features',
    test_battery_id=8,
    batch_size=128,
    normalized_type='minmax',
    minmax_range=(-1,1),
    batch=1, model='CNN',
    lr=2e-3, weight_decay=5e-4,
    n_epoch=39, early_stop=30,
    device='cpu', save_folder='results',
    experiment_num=1
)

d = XJTUDdataset(args).get_features(test_battery_id=8)
m = SOHMode(args)

p = "results/XJTU-handcraft_features/CNN_BASELINE_67/batch8-testbattery8/experiment1"

m.Train(d["train"], d["valid"], d["test"], save_folder=p)

r = np.load(f"{p}/CNN_handcraft_features_results.npz")
print("Battery 8:", r["test_errors"])
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

base = "results/XJTU-handcraft_features/CNN_BASELINE_67"
fs   = "results/XJTU-handcraft_features/CNN_FS40_CORRECT"

rows = []

for b in range(2, 9):
    r1 = np.load(f"{base}/batch{b}-testbattery{b}/experiment1/CNN_handcraft_features_results.npz")
    r2 = np.load(f"{fs}/batch{b}-testbattery{b}/experiment1/CNN_handcraft_features_results.npz")

    rows.append([
        b, *r1["test_errors"], *r2["test_errors"]
    ])

df = pd.DataFrame(rows, columns=[
    "Battery",
    "Baseline_MAE","Baseline_MAPE","Baseline_MSE","Baseline_RMSE",
    "Top40_MAE","Top40_MAPE","Top40_MSE","Top40_RMSE"
])

print("=== BATTERY-WISE COMPARISON ===")
display(df.round(6))

avg = df.iloc[:,1:].mean()
print("\n=== AVERAGE RESULTS ===")
display(avg.round(6))

print("\n=== IMPROVEMENT (%) ===")
for m in ["MAE","MAPE","MSE","RMSE"]:
    imp = (avg[f"Baseline_{m}"]-avg[f"Top40_{m}"]) / avg[f"Baseline_{m}"] * 100
    print(f"{m}: {imp:.2f}%")

# Plot MAE
x = np.arange(len(df))
w = 0.35

plt.figure(figsize=(9,5))
plt.bar(x-w/2, df["Baseline_MAE"], w, label="67 Features")
plt.bar(x+w/2, df["Top40_MAE"], w, label="Top-40 Features")
plt.xticks(x, df["Battery"])
plt.xlabel("Test Battery")
plt.ylabel("MAE")
plt.title("67 Features vs Top-40 Features")
plt.legend()
plt.tight_layout()
plt.show()
```
